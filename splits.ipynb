```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
for j in range(1,101):
    df = pd.read_csv(f'data/staminadata/{j}_training.txt.dat', sep=';')
    df.columns = [col.strip() for col in df.columns]
    vals = list(df.columns)[0].split()
    sample_size = int(vals[0])
    alphabet_size = int(vals[1])
    series = pd.Series(df.values.flatten())
    unique_traces = pd.unique(series)
    y = []
    x = []
    for val in unique_traces:
        y.append(val[0])
        x.append(val[2:])
    x_train, x_test, y_train, y_test = train_test_split(
        x, y,
        test_size=0.2,
        stratify=y,
        random_state=10
    )
    with open(f"data/stamina_split/{j}_training.txt.dat", 'w') as file:
        file.write(f"{len(x_train)} {alphabet_size}\n")
        for i, sample in enumerate(x_train):
            file.write(f"{y_train[i]} {sample}\n")
    with open(f"data/stamina_split/{j}_test.txt.dat", 'w') as file:
        file.write(f"{len(x_test)} {alphabet_size}\n")
        for i, sample in enumerate(x_test):
            file.write(f"{y_test[i]} {sample}\n")
```
